## CLASSES IN PYTHON 

## Creating a Class and Object

```python
class Dog:
    def __init__(self, name, breed):
        self.name = name
        self.breed = breed
    def bark(self):
        print(f"{self.name} barks!")

# Create an object of the Dog class
my_dog = Dog("Buddy", "Golden Retriever")
# Access attributes and call methods
print(my_dog.name)  
my_dog.bark()  
```

## Inheritance

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def eat(self):
        print(f"{self.name} is eating.")

class Dog(Animal):
    def bark(self):
        print(f"{self.name} barks!")

my_dog = Dog("Buddy")
my_dog.eat()  
my_dog.bark() 
```

## Encapsulation

```python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private attribute
    def deposit(self, amount):
        self.__balance += amount
    def withdraw(self, amount):
        if self.__balance >= amount:
            self.__balance -= amount
        else:
            print("Insufficient funds.")
    def get_balance(self):
        return self.__balance
```

## 3. Create a class Circle with attributes radius and methods to calculate area and circumference.

```python
import math
class Circle:
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius ** 2
    def circumference(self):
        return 2 * math.pi * self.radius

# Example
circle1 = Circle(5)
print("Area:", circle1.area())
print("Circumference:", circle1.circumference())
```

## 4. Create a class Rectangle with attributes length and width and methods to calculate area and perimeter.

```python
class Rectangle:
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width
    def perimeter(self):
        return 2 * (self.length + self.width)


# Example
r1 = Rectangle(10, 4)
print("Area:", r1.area())
print("Perimeter:", r1.perimeter())
```

## 5. Create a class Student with attributes name, roll_number, and marks. Implement a method to calculate the average marks.

```python
class Student:
    def __init__(self, name, roll_number, marks):
        self.name = name
        self.roll_number = roll_number
        self.marks = marks  
    def calculate_average(self):
        return sum(self.marks) / len(self.marks)


# Example
s1 = Student("Wali", "SP26-BBD-047", [80, 85, 90])
print("Average:", s1.calculate_average())
```

## 8. Create a class Bank with attributes name, account_number, and balance. Implement methods to deposit and withdraw money.

```python
class Bank:
    def __init__(self, name, account_number, balance):
        self.name = name
        self.account_number = account_number
        self.balance = balance
    def deposit(self, amount):
        self.balance += amount
        print("Deposit Successful")
    def withdraw(self, amount):
        if amount <= self.balance:
            self.balance -= amount
            print("Withdrawal Successful")
        else:
            print("Insufficient Balance")

# Example
account1 = Bank("Wali", "123456", 100000)
account1.deposit(2000)
account1.withdraw(50000)
print("Current Balance:", account1.balance)
```

## 9. Create a class Shape with a method to calculate area. Create subclasses Circle, Rectangle, and Triangle that inherit from Shape and implement their specific area calculations.

```python
import math
class Shape:
    def area(self):
        pass
        
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius
    def area(self):
        return math.pi * self.radius ** 2

class Rectangle(Shape):
    def __init__(self, length, width):
        self.length = length
        self.width = width
    def area(self):
        return self.length * self.width


class Triangle(Shape):
    def __init__(self, base, height):
        self.base = base
        self.height = height
    def area(self):
        return 0.5 * self.base * self.height


# Example
shapes = [Circle(4), Rectangle(5, 6), Triangle(10, 3)]
for shape in shapes:
    print("Area:", shape.area())
```

## 10. Create a class Animal with attributes name and sound. Create subclasses Dog, Cat, and Cow that inherit from Animal and implement their specific sounds

```python
class Animal:
    def __init__(self, name):
        self.name = name
    def make_sound(self):
        pass

class Dog(Animal):
    def make_sound(self):
        return "Woof"

class Cat(Animal):
    def make_sound(self):
        return "Meow"

class Cow(Animal):
    def make_sound(self):
        return "Moo"

# Example
animals = [Dog("Rockson"), Cat("Bella"), Cow("Max")]
for animal in animals:
    print(animal.name, "says", animal.make_sound())
```
